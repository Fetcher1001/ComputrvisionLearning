```python
import cv2 as cv
import numpy as np

path = 'C:\\Users\\johan\\computervision Learning\\ComputrvisionLearning\\Experimentation\\pictures\\Polygone.png'
gray =cv.imread(path, cv.COLOR_BAYER_BG2GRAY)
blurred = cv.GaussianBlur(gray, (5,5),0)


edged = cv.Canny(blurred, 30,150)
contours, _ = cv.findContours(edged.copy(), cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

print(f'There are {len(contours)} contours in the image')

cv.drawContours(gray, contours, -1, (0,255,0), 2)
cv.imshow('Contours', gray)
cv.waitKey(0)
cv.destroyAllWindows()

for cnt in contours:
    peri = cv.arcLength(cnt, True)
    approx = cv.approxPolyDP(cnt, 0.04 * peri, True)
```
