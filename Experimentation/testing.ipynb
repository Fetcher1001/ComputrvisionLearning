```python
import cv2 as cv
import numpy as np
print(cv.__version__)

test_one = 'C:\\Users\\johan\\computervision Learning\\ComputrvisionLearning\\Experimentation\\pictures\\Polygone.png'
test_two = 'C:\\Users\\johan\\computervision Learning\\ComputrvisionLearning\\Experimentation\\pictures\\52b89327e0c751cb5f3c48aaba3d3ca1.jpg'
# import a picture 
img = cv.imread(test_one)
#cv.imshow('Original Image', img)
#cv.waitKey(0)

print("image shape",img.shape)
print("pixel",img[0,0])

new_img = img
new_img[:,:,1] = 0
cv.imshow('new Image', new_img)
cv.waitKey(0)

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
cv.imshow('Gray Image -win anti old way mode', gray)
cv.waitKey(0)

edges = cv.Canny(img,100,200)
cv.imshow('Canny Edges', edges)

kernel = np.ones((3,3),np.uint8)
thickedges = cv.dilate(edges,kernel, iterations=2)
binary_img = thickedges.copy()
data = cv.imwrite('thick_edges.jpg', thickedges)
contours, hierarchy = cv.findContours(edges, cv.RETR_TREE, cv.CHAIN_APPROX_SIMPLE)

for i, contour in enumerate(contours):
    print(f'Contour {i}:')
    print(contour)
    print(len(contour))

print(f'Number of contours found: {len(contours)}')
draw_img = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

cv.drawContours(draw_img, contours, -1, (0,255,0), 2)
cv.imshow('Contours', draw_img)
cv.imshow('Thick Edges', thickedges)
cv.waitKey(0)
cv.destroyAllWindows()

```
